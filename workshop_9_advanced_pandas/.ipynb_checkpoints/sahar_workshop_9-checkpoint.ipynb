```python
#Product Analysis
```

```python
# Group products by category and calculate average price for each category
products['category']
```

```python
products.groupby('category')
```

```python
products.groupby('category')['price']
#pandas look at price inside each category
```

```python
average_price_by_category = products.groupby('category')['price'].mean().round(2)
```

```python
average_price_by_category
```

```python
average_price_by_category.apply(lambda x: f"${x:.2f}")
```

```python
most_expensive_category = average_price_by_category.idxmax()
```

```python
print("Most Expensive Category is", most_expensive_category)
```

```python
# most revenue generated items:
product_revenue = sales.groupby('product_id')['total_amount'].sum()

most_product_id = product_revenue.idxmax()

products.loc[products['product_id'] == most_product_id, 'product_name']
```

```python
# least revenue generated item
product_revenue = sales.groupby('product_id')['total_amount'].sum()

least_product_id = product_revenue.idxmin()

products.loc[products['product_id'] == least_product_id, 'product_name']
```

```python
average_price_by_category = products.groupby('category')['price'].mean().round(2).sort_values(ascending=False)
```

```python
average_price_by_category 
```

```python
average_price_by_category.apply(lambda x: f"${x:.2f}")
```

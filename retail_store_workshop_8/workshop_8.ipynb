```python
#Part 2: Data Cleaning
#Handling Missing Values
```

```python
#Identify all missing values in each dataset
# it is already answered in "Check for missing values in all dataframes" part
```

```python
#For numerical columns with missing values, replace them with the column mean

inventory_df['reorder_level'] .mean()

```

```python
inventory_df['reorder_level'] = inventory_df['reorder_level']. fillna(inventory_df['reorder_level'].mean())
inventory_df
```

```python
stores_df['size_sqft'].mean()
```

```python
stores_df['size_sqft'] = stores_df['size_sqft']. fillna(stores_df['size_sqft'].mean())
stores_df
```

```python
products_df['weight'].mean()
```

```python
products_df['weight'] = products_df['weight']. fillna(products_df['weight'].mean())
products_df
```

```python
#For categorical columns with missing values, replace them with the most frequent value
customers_df = customers_df.fillna(customers_df.mode().iloc[0])
```

```python
customers_df
```

```python
stores_df = stores_df.fillna(stores_df.mode().iloc[0])
```

```python
stores_df
```

```python
inventory_df = inventory_df.fillna(inventory_df.mode().iloc[0])
```

```python
inventory_df
```

```python
sales_df = sales_df.fillna(sales_df.mode().iloc[0])
```

```python
sales_df
```

```python
products_df = products_df.fillna(products_df.mode().iloc[0])
```

```python
products_df
```
